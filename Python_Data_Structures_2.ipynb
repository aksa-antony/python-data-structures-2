```python
#Age list
age_list = [20,24, 25, 26, 27, 28,29,30]
print(age_list)
```

```python
#Name list
name_list = ["Aksa", "Anagha", "Dona", "Ancy", "Hanna"]
print(name_list)
```

```python
#List operations
#Append
name_list.append("Yazhini")
print(name_list)
```

```python
#insert
age_list.insert(2, 30)
print(age_list)
```

```python
#remove
name_list.remove("Yazhini")
print(name_list)
```

```python
#pop
age_list.pop()
print(age_list)
```

```python
#extend
age_list.extend([29, 30, 26])
print(age_list)
```

```python
#sort
age_list.sort(reverse=True)
print(age_list)
```

```python
#max,min,sum
age_list = [20,24, 25, 26, 27, 28,29,30]
print(max(age_list))
print(min(age_list))
print(sum(age_list))
```

```python
#access & print specific
student_marks={
  "Aksa":78,
  "Hanna":80,
  "Anagha":70,
  "Dona":79,
  "Ancy":81}
print(student_marks["Hanna"])
```

```python
#create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print(set1)
print(set2)
```
